```python
import cv2# Tem métodos para realizar operações em imagens
import numpy as np# Para manipular arrays
from google.colab.patches import cv2_imshow# Exibe uma imagem
import matplotlib.pyplot as plt# Exibir figuras
```

```python
def plot_img_plt(img, titulo, width, height):
  plt.figure(figsize=(width, height))
  plt.plot()
  plt.title(titulo)
  plt.imshow(img)
```

# Operações básicas: negativo de uma imagem

```python
img_baboon = cv2.imread("./pdsi/imagens/baboon.png", cv2.IMREAD_COLOR)
#cv2_imshow(img_baboon)
img_baboon_rgb = cv2.cvtColor(img_baboon, cv2.COLOR_BGR2RGB)

plot_img_plt(img_baboon_rgb, "Baboon RGB", 5, 5)
```

```python
width, height, channel = img_baboon.shape
img_negativa = np.full(shape=(width, height, channel), fill_value=0)
```

```python
for x in range(0, img_baboon_rgb.shape[0]):
  for y in range(0, img_baboon_rgb.shape[1]):
    img_negativa[x][y] = 255 - img_baboon[x][y]

def negative_img(img):
  width, height, channels = img.shape
  img_neg = np.full(shape=(width, height, channels), fill_value=0)

  for x in range(0, img.shape[0]):
    for y in range(0, img.shape[1]):
      img_neg[x][y] = 255 - img[x][y]

  return img_neg
```

```python
plt.figure()

plt.subplot(1, 2, 1)
plt.title("Baboon Original")
plt.plot()
plt.imshow(img_baboon_rgb)

plt.subplot(1, 2, 2)
plt.title("Baboon Negativo")
plt.plot()
plt.imshow(img_negativa)
```

# Translação

```python
width, height, channels = img_baboon_rgb.shape
width_4, height_4 = width / 4, height / 4
```

```python
matriz_translacao = np.float32([[1, 0, width_4], [0, 1, height_4]])
img_transladada = cv2.warpAffine(img_baboon_rgb, matriz_translacao,( width, height))

plt.figure()

plt.subplot(1, 2, 1)
plt.plot()
plt.title("Baboon Original")
plt.imshow(img_baboon_rgb)

plt.subplot(1, 2, 2)
plt.plot()
plt.title("Baboon Transladado")
plt.imshow(img_transladada)
```

# Reflexão (Espelhamento)

```python
img_lena = cv2.imread("./pdsi/imagens/lena.png", cv2.IMREAD_COLOR)
img_lena_rgb = cv2.cvtColor(img_lena, cv2.COLOR_BGR2RGB)

img_lenaX = np.flip(img_lena_rgb, axis=1)
img_lenaY = np.flip(img_lena_rgb, axis=0)
img_lenaXY = np.flip(img_lena_rgb, axis=(0, 1))

plt.figure(figsize=(15, 15))

#Imagem original
plt.subplot(2, 2, 1)
plt.plot()
plt.title("Imagem Original")
plt.imshow(img_lena_rgb)

plt.subplot(2, 2, 2)
plt.plot()
plt.title("Imagem Espelhada no eixo X")
plt.imshow(img_lenaX)

plt.subplot(2, 2, 3)
plt.plot()
plt.title("Imagem Original")
plt.imshow(img_lenaY)

plt.subplot(2, 2, 4)
plt.plot()
plt.title("Imagem Original")
plt.imshow(img_lenaXY)
```

# Resize

```python
img_cam = cv2.imread("./pdsi/imagens/cameraman.png", cv2.IMREAD_COLOR)
img_cam_rgb = cv2.cvtColor(img_cam, cv2.COLOR_BGR2RGB)
img_cam_rgb.shape
```

```python
Fx, Fy = 1.5, 1.25

img_cam_resized = cv2.resize(img_cam_rgb, None, fx=Fx, fy=Fy, interpolation=cv2.INTER_CUBIC)

plt.figure()

plt.subplot(1, 2, 1)
plt.plot()
plt.title("Img Original")
plt.imshow(img_cam_rgb)

plt.subplot(1, 2, 2)
plt.plot()
plt.title("Img Resized")
plt.imshow(img_cam_resized)
```
