# Lendo imagem usando OpenCV

```python
import cv2# Várias operações pra realizar em imagens
import numpy as np# Converter as imagens em array
import matplotlib.pyplot as plt# Plotar imagens
from google.colab.patches import cv2_imshow# Plotar imagens
```

```python
path = "./pdsi/imagens/baboon.png"
img_bgr = cv2.imread(path, cv2.IMREAD_COLOR_BGR)
cv2_imshow(img_bgr)
```

```python
# Usando matplotlib
def plot_img_plt(img, titulo, width, height):
  plt.figure(figsize=(width, height))
  plt.plot()
  plt.title(titulo)
  plt.imshow(img)
```

```python
plot_img_plt(img_bgr, "Baboon", 5, 5)
```

```python
# Convertendo BGR para RGB
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plot_img_plt(img_rgb, "Baboon RGB", 5, 5)
```

```python
print(f"Formato da imagem: {img_rgb.shape} - {type(img_bgr)}")
```

```python
# Pegando cada canal separado usando manualmente
canal_blue = img_bgr[:, :, 0]
canal_green = img_bgr[: , :, 1]
canal_red = img_bgr[:, :, 2]

plot_img_plt(canal_blue, "Baboon Blue Chanel", 5, 5)
```

```python
plot_img_plt(canal_green, "Baboon Green Chanel", 5, 5)
```

```python
plot_img_plt(canal_red, "Baboon Red Chanel", 5, 5)
```

```python
# Usando opencv
canal_blue, canal_green, canal_red = cv2.split(img_bgr)
cv2_imshow(canal_blue)
```

```python
cv2_imshow(canal_green)
```

```python
cv2_imshow(canal_red)
```

```python
img_grayscale = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
cv2_imshow(img_grayscale)
```

```python
for i in range(0, img_grayscale.shape[0]):
  for j in range(0, img_grayscale.shape[1]):
    print(f"Valor de cinza da coordenada {i, j}: {img_grayscale[i, j]}")
```
